```python
import time
import psutil
```

```python
#加载别人的模型
import gensim.models.keyedvectors as word2vec
start_time = time.time()
initial_memory = psutil.Process().memory_info().rss / (1024 * 1024)  # Convert to MB

#这个模型可以在https://pan.baidu.com/s/1Rn7LtTH0n7SHyHPfjRHbkg这里下载到，解压后（1.8GB左右）放到和程序相同文件夹
#该模型使用百度百科进行训练，维度为300维，详见https://github.com/Embedding/Chinese-Word-Vectors?spm=a2c22.12282016.0.0.26395a20svQxNt

model_baike = word2vec.KeyedVectors.load_word2vec_format('sgns.financial1.bigram-char',binary=False, encoding="utf8",  unicode_errors='ignore')
print('loading finished!')
end_time = time.time()
final_memory = psutil.Process().memory_info().rss / (1024 * 1024)  # Convert to MB

# 输出加载所需时间和内存使用情况
print(f"百度百科300维模型 Loading time: {end_time - start_time} seconds")
print(f"百度百科300维模型 Memory usage: From {initial_memory} MB to {final_memory} MB")
```

```python
#打印向量
print(model_baike['安全'])
```

```python
def similar_words_bk(in_word,topk=5):
    similar_words = model_baike.most_similar(in_word, topn=topk)
    print("金融模型，与 "+in_word+" 最相似的词是：")

    for word, similarity in similar_words:
        print(f"{word}: {similarity:.4f}")
   
```

```python
similar_words_bk('泄露',10)
#在只有keyedvectors的情况（大部分可供下载的预训练word2vec的提供方式）下，不能给模型加入新的词
```

```python
similar_words_bk('数据',10)
```

```python
similar_words_bk('风险',10)
```

```python
similar_words_bk('防火墙',10)
```

```python
similar_words_bk('AI',10)
```

```python
result = model_baike.most_similar(positive=["金融", "安全"], negative=["数据"], topn=10)
print("“金融”-“数据”+“安全”对应的词是：")
for word, similarity in result:
    print(f"{word}: {similarity:.4f}")
```
