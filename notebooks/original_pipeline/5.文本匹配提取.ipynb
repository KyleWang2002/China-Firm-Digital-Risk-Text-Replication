```python
def extract_keyword_lines(MADdir):
    results = []
    
    for filename in os.listdir(MADdir):
        if filename.endswith(".txt"):
            filepath = os.path.join(MADdir, filename)
            with open(filepath, encoding='utf-8') as f:
                for line_num, line in enumerate(f, 1):
                    if pattern.search(line):
                        # 移除换行符和多余空格
                        clean_line = line.strip()
                        results.append({
                            "文件名": filename,
                            "行号": line_num,
                            "内容": clean_line
                        })
    return results
```

```python
# 4. 主执行流程
if __name__ == "__main__":
    # 配置参数
    MAD_DIR = "C:/Users/Lenovo/Desktop/上市公司文本处理/项目代码/年报文件/2024/txt年报/MDA_cleaned"  # 修改为实际路径
    OUTPUT_FILE = "C:/Users/Lenovo/Desktop/上市公司文本处理/项目代码/年报文件/2024/txt年报/MDA_cleaned/提取结果.xlsx"
    
    # 执行提取
    matched_data = extract_keyword_lines(MAD_DIR)
    
    # 转换为DataFrame并保存Excel
    df = pd.DataFrame(matched_data)
    df.to_excel(OUTPUT_FILE, index=False)
    print(f"成功提取{len(matched_data)}行数据到 {OUTPUT_FILE}")
```
